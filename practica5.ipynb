## 1. SINTAXIS BÁSICA Y COMPARACIÓN CON DEF

Una función tradicional de una sola línea:

```python
def calcular_iva(precio):
    return precio * 0.13
```

La misma lógica resumida en una expresión lambda

```python
calcular_iva_lambda = lambda precio: precio * 0.13
```

```python
print("IVA normal:", calcular_iva(100))
print("IVA lambda:", calcular_iva_lambda(100))
```

## 2. MÚLTIPLES ARGUMENTOS

Operación matemática directa con varios parámetros

```python
area_rectangulo = lambda base, altura: base * altura
distancia_total = lambda velocidad, tiempo: velocidad * tiempo
```

```python
print("Área:", area_rectangulo(5, 8))
print("Distancia:", distancia_total(60, 2.5), "km")
```

## 3. CONDICIONALES EN LÍNEA (OPERADOR TERNARIO)

Sintaxis: valor_si_verdadero if condicion else valor_si_falso

```python
evaluar_estado = lambda nota: "Aprobado" if nota >= 6.0 else "Reprobado"
es_par = lambda n: True if n % 2 == 0 else False
```

```python
print("Resultado 1:", evaluar_estado(7.5))
print("Resultado 2:", evaluar_estado(4.0))
print("¿El 10 es par?:", es_par(10))
```

## 4. APLICACIÓN RÁPIDA DENTRO DE CICLOS O COMPRENSIONES

Formateo de texto en bucles cotidianos

```python
formatear_codigo = lambda prefijo, numero: f"{prefijo}-{numero:04d}"

ids_generados = [formatear_codigo("USR", i) for i in range(1, 6)]
print("Códigos generados:", ids_generados)
```

### 5.1 Refactorización de funciones convencionales

Tres funciones tradicionales con `def` convertidas a `lambda`

```python
def aplicar_descuento_a_precio(precio_original: float, porcentaje_descuento: float) -> float:
    return precio_original - (precio_original * porcentaje_descuento / 100)

def convertir_celsius_a_fahrenheit(grados_celsius: float) -> float:
    return (grados_celsius * 9/5) + 32

def normalizar_nombre_completo(primer_nombre: str, primer_apellido: str) -> str:
    return f"{primer_nombre.strip().capitalize()} {primer_apellido.strip().capitalize()}"


obtener_precio_con_descuento = lambda precio_original, porcentaje_descuento: precio_original - (precio_original * porcentaje_descuento / 100)
transformar_celsius_a_fahrenheit = lambda grados_celsius: (grados_celsius * 9/5) + 32
estandarizar_nombre_completo = lambda primer_nombre, primer_apellido: f"{primer_nombre.strip().capitalize()} {primer_apellido.strip().capitalize()}"


monto_base = 200.0
porcentaje_aplicado = 15.0
temperatura_celsius = 25.0
nombre_usuario = "ana"
apellido_usuario = "lopez"

precio_final_estandar = aplicar_descuento_a_precio(monto_base, porcentaje_aplicado)
precio_final_lambda = obtener_precio_con_descuento(monto_base, porcentaje_aplicado)

print(f"Precio con descuento (Función): ${precio_final_estandar:.2f}")
print(f"Precio con descuento (Lambda):  ${precio_final_lambda:.2f}\n")

fahrenheit_estandar = convertir_celsius_a_fahrenheit(temperatura_celsius)
fahrenheit_lambda = transformar_celsius_a_fahrenheit(temperatura_celsius)

print(f"Temperatura convertida (Función): {temperatura_celsius}°C equivale a {fahrenheit_estandar}°F")
print(f"Temperatura convertida (Lambda):  {temperatura_celsius}°C equivale a {fahrenheit_lambda}°F\n")


cliente_estandar = normalizar_nombre_completo(nombre_usuario, apellido_usuario)
cliente_lambda = estandarizar_nombre_completo(nombre_usuario, apellido_usuario)

print(f"Nombre formateado (Función): {cliente_estandar}")
print(f"Nombre formateado (Lambda):  {cliente_lambda}")
```

### 5.2 Validaciones con operadores ternarios

Expresiones lambda que devuelven respuestas booleanas o de estado

```python
clasificar_signo_numero = lambda valor_numerico: (
    "Positivo" if valor_numerico > 0 
    else "Negativo" if valor_numerico < 0 
    else "Cero"
)

es_persona_mayor_de_edad = lambda edad_anios: edad_anios >= 18


numero_positivo = 15
numero_negativo = -8
numero_neutro = 0

edad_menor = 17
edad_limite = 18


signo_pos = clasificar_signo_numero(numero_positivo)
signo_neg = clasificar_signo_numero(numero_negativo)
signo_cero = clasificar_signo_numero(numero_neutro)

print(f"El signo de {numero_positivo} es: {signo_pos}")
print(f"El signo de {numero_negativo} es: {signo_neg}")
print(f"El signo de {numero_neutro} es: {signo_cero}\n")

resultado_edad_menor = "Sí" if es_persona_mayor_de_edad(edad_menor) else "No"
resultado_edad_limite = "Sí" if es_persona_mayor_de_edad(edad_limite) else "No"

print(f"¿Una persona con {edad_menor} años es mayor de edad?: {resultado_edad_menor} ({es_persona_mayor_de_edad(edad_menor)})")
print(f"¿Una persona con {edad_limite} años es mayor de edad?: {resultado_edad_limite} ({es_persona_mayor_de_edad(edad_limite)})")
```

### 5.3 Procesamiento en lote

Aplicación de expresiones lambda dentro de un bucle sobre una colección de valores numéricos

```python
calcular_precio_con_recargo = lambda precio_base: precio_base * 1.10

lista_precios_originales = [50, 120, 35.5, 200, 78.9]
lista_precios_con_recargo = [calcular_precio_con_recargo(precio) for precio in lista_precios_originales]

print(f"Precios originales: {lista_precios_originales}")
print(f"Precios con recargo del 10%: {lista_precios_con_recargo}\n")

convertir_moneda_local_a_dolares = lambda monto_moneda_local: monto_moneda_local / 8.75

lista_montos_originales = [100, 250, 500, 1000]
lista_montos_en_dolares = [round(convertir_moneda_local_a_dolares(monto), 2) for monto in lista_montos_originales]

print(f"Montos en moneda local: {lista_montos_originales}")
print(f"Montos convertidos a dólares: {lista_montos_en_dolares}")
```
