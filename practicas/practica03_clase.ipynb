# Planteamiento de la Práctica 3: Métodos de Listas, Sets y sus Patrones de Uso (LIFO, FIFO y Operaciones de Conjuntos)

## 1. Métodos Esenciales de Listas

```python
# Inserción, búsqueda y ordenamiento
frutas = ["manzana", "banana", "cereza"]

frutas.append("durazno")
frutas.insert(1, "mango")
print("Lista tras inserciones:", frutas)

frutas.extend(["uva", "kiwi"])
print("Lista extendida:", frutas)

posicion = frutas.index("cereza")
print("Índice de 'cereza':", posicion)

frutas.sort()
print("Lista ordenada alfabéticamente:", frutas)

frutas.reverse()
print("Lista invertida:", frutas)
```

## 2. Patrón LIFO (Pila / Stack) con Listas

```python
# Simulación de historial de navegación web
historial = []

# Visitar páginas (Push)
historial.append("google.com")
historial.append("github.com")
historial.append("stackoverflow.com")
print("Historial actual:", historial)

# Botón atrás (Pop)
pagina_actual = historial.pop()
print("Saliendo de:", pagina_actual)
print("Página donde queda el usuario:", historial[-1])
```

## 3. Patrón FIFO (Cola / Queue)

```python
from collections import deque

# Simulación de cola de impresión de documentos
cola_impresion = deque()

# Agregar documentos a la cola (Enqueue)
cola_impresion.append("reporte.pdf")
cola_impresion.append("factura.pdf")
cola_impresion.append("presentacion.pptx")
print("Documentos en cola:", list(cola_impresion))

# Procesar documentos en orden de llegada (Dequeue)
doc_procesado = cola_impresion.popleft()
print("Imprimiendo:", doc_procesado)
print("Siguiente en fila:", cola_impresion[0])
```

## Métodos Principales de Sets y Operaciones Mutables

```python
# Creación, adición y eliminación
usuarios_activos = {"alice", "bob", "charlie"}

usuarios_activos.add("david")
# Intentar agregar un duplicado no altera el conjunto
usuarios_activos.add("alice")

# Eliminación segura (discard no lanza KeyError si no existe)
usuarios_activos.discard("bob")
usuarios_activos.discard("usuario_inexistente")

print("Usuarios activos finales:", usuarios_activos)
```

## 5. Caso de Uso Práctico de Sets: Control de Permisos y Ajustes de Roles

```python
# Permisos requeridos para ejecutar un módulo
permisos_requeridos = {"leer", "escribir", "ejecutar", "eliminar"}

# Permisos asignados a dos roles de usuario
permisos_editor = {"leer", "escribir"}
permisos_admin = {"leer", "escribir", "ejecutar", "eliminar", "auditar"}

# Verificar si un rol posee todos los permisos necesarios (Subset)
es_suficiente = permisos_requeridos.issubset(permisos_editor)
print("¿Editor puede ejecutar el módulo?:", es_suficiente)

# Determinar qué permisos le faltan al editor (Diferencia)
permisos_faltantes = permisos_requeridos - permisos_editor
print("Permisos faltantes para el editor:", permisos_faltantes)

# Permisos comunes entre editor y admin (Intersección)
permisos_comunes = permisos_editor & permisos_admin
print("Permisos compartidos:", permisos_comunes)
```
