Caso 1 (LIFO o FIFO): Diseñe una solución basada en el patrón seleccionado (ejemplo: sistema deshacer/rehacer, atención a clientes en un banco, cola de reproducción musical o gestión de tareas pendientes).

```python
# Cola de reproducción musical

# importamos deque from collections
from collections import deque

# Usamos deque porque funciona bien para crear una cola.

# Justificaión

# deque es una estructura de datos de Python que funciona como una cola de elementos.

cola_de_musica = deque()

# Agregamos canciones al final de la cola.

cola_de_musica.append("Stressed out")
cola_de_musica.append("24k magic")
cola_de_musica.append("Safe and sound")

# Mostramos las canciones en espera.

print("Canciones en cola:", list(cola_de_musica))

# Sacamos la primera canción de la cola para reproducirla.

cancion_actual = cola_de_musica.popleft()

print("Reproduciendo:", cancion_actual)

# Mostramos la siguiente canción.

print("Siguiente en fila:", cola_de_musica)
```

```python
# Filtro de etiquetas duplicadas

# Lista de etiquetas, incluyendo algunas repetidas

etiquetas = {"rock", "pop", "musica", "rock", "concierto"}

# Los conjuntos eliminan automáticamente los elementos duplicados

print("Etiquetas únicas:", etiquetas)

# Agregamos una nueva etiqueta

etiquetas.add("jazz")

# Intentamos agregar una etiqueta repetida

# No se duplicará porque los sets no permiten duplicados

etiquetas.add("rock")

# Eliminamos una etiqueta de forma segura

etiquetas.discard("concierto")

print("Etiquetas finales:", etiquetas)
```
