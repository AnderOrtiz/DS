## Sistema Integral de Caja y Despacho

Implemente las siguientes funcionalidades obligatorias en su programa de simulación:

- Fila en Caja (FIFO): Atender clientes en estricto orden de llegada usando collections.deque con popleft(). Validar qué sucede cuando se intenta cobrar y no hay nadie en la fila.

- Bolsa de Productos (LIFO): Cada cliente tiene una bolsa (pila). Al momento de pagar, los productos se ingresan en un orden, pero el cajero debe desempacar o registrar primero el último producto guardado (pop()).

- Módulos de Cupones y Membresías (Sets): Manejar un conjunto de códigos de membresía vigentes. Registrar el cupón usado para evitar que se repita en la misma sesión.

```python
from collections import deque


fila = deque()

fila.append({
    "nombre": "Anderson",
    "productos": ["Pan", "Leche", "Huevos"]
})

fila.append({
    "nombre": "Carlos",
    "productos": ["Arroz", "Frijoles", "Jugo"]
})

fila.append({
    "nombre": "Maria",
    "productos": ["Cafe", "Azucar"]
})



membresias_vigentes = {
    "MEM001",
    "MEM002",
    "MEM003"
}

cupones_usados = set()


def atender_cliente(fila):

    if not fila:
        print("No hay nadie en la fila.")
        return

    cliente_actual = fila.popleft()

    print(f"\nAtendiendo a: {cliente_actual['nombre']}")

    cobrar_cliente(cliente_actual)



def cobrar_cliente(cliente):

    print("Registrando productos:")

    bolsa = cliente["productos"]

    while bolsa:
        producto = bolsa.pop()
        print(f"  Registrando: {producto}")

    print("Todos los productos fueron registrados.")



def validar_membresia(codigo):

    if codigo in membresias_vigentes:
        print("Membresía válida.")
        return True

    print("Membresía no válida.")
    return False



def usar_cupon(codigo):

    if codigo in cupones_usados:
        print("El cupón ya fue utilizado.")
        return False

    cupones_usados.add(codigo)

    print("Cupón aplicado correctamente.")
    return True



print("========== CAJA ==========")

for i in range(4):
    atender_cliente(fila)


print("\n========== MEMBRESÍAS ==========")

validar_membresia("MEM001")
validar_membresia("MEM999")


print("\n========== CUPONES ==========")

usar_cupon("DESC10")
usar_cupon("DESC20")
usar_cupon("DESC10")
```
