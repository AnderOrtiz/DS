# Justificación del uso de Listas, Tuplas y Sets

En este miniprograma se utiliza la temática de **música** para representar el uso de los tipos de datos **listas, tuplas y sets**.

Se eligió cada tipo de dato de acuerdo con la operación que se realizará:

* **Lista:** para almacenar y modificar canciones.
* **Tupla:** para guardar información que no necesita cambiar.
* **Set:** para almacenar géneros musicales sin duplicados.


```python
# Lista de reproducción musical
# Las listas permiten almacenar y modificar elementos.

canciones = ["Imagine", "Thriller", "Bohemian Rhapsody", "Imagine"]

print("Lista de reproducción:")
print(canciones)

# Agregamos una nueva canción
canciones.append("Billie Jean")

print("\nDespués de agregar una canción:")
print(canciones)


# Tupla con información de una canción
# Las tuplas permiten guardar datos que no necesitamos modificar.

cancion = ("Bohemian Rhapsody", "Queen", 1975)

print("\nInformación de la canción:")
print("Título:", cancion[0])
print("Artista:", cancion[1])
print("Año:", cancion[2])


# Set de géneros musicales
# Los sets no permiten elementos repetidos.

generos = {"Rock", "Pop", "Rock", "Jazz"}

print("\nGéneros musicales sin repetir:")
print(generos)

# Agregamos un nuevo género
generos.add("Reggae")

print("\nGéneros finales:")
print(generos)
```
