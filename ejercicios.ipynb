### Ejercicio 1: Árbol Genealógico Familiar
Utilizando la clase Nodo, modele un árbol genealógico de 3 niveles con al menos 7 personas:

- Coloque en la raíz al abuelo/abuela.
- Asigne en los enlaces izquierdo y derecho a dos hijos (padres/tíos).
- Asigne a cada hijo al menos dos descendientes (nietos) en sus respectivos enlaces izquierdo y derecho.
- Imprima las relaciones de forma manual accediendo a los atributos para comprobar que cada nodo apunta al familiar correcto.

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

raiz = Nodo("Abuelo")
raiz.izq = Nodo("Papá")
raiz.der = Nodo("Tío")

raiz.izq.izq = Nodo("Hermano")
raiz.der.der = Nodo("Primo")

print(f"La raiz del nodo es mi {raiz.valor}")
print(f"El nodo izquierdo de la raiz es mi {raiz.izq.valor}")
print(f"El nodo derecho de la raiz es mi {raiz.der.valor}")
print(f"El nodo derecho del nodo derecho de la raiz es mi {raiz.der.der.valor}")
print(f"El nodo izquierdo del nodo izquierdo de la raiz es mi {raiz.izq.izq.valor}")
```

### Ejercicio 2: Menú Interactivo de Decisiones
Cree un árbol de toma de decisiones para un sistema de soporte técnico o menú de opciones:

- Construya un árbol donde la raíz sea una pregunta inicial y los enlaces izquierdo y derecho lleven a opciones como "Sí" y "No" con sus respectivas acciones finales.
- Ejecute e imprima el recorrido en Preorden para listar el orden en el que se evaluarían las opciones desde la raíz hacia las hojas.
- Ejecute e imprima el recorrido en Postorden para simular el cierre de los procesos de soporte desde las hojas hacia la raíz.

```python
raiz = Nodo("¿El router tiene las luces encendidas?")

raiz.izq = Nodo("¿Puedes navegar en algún sitio web?")
raiz.izq.izq = Nodo("[Solución] La conexión está funcionando correctamente.")  # Sí
raiz.izq.der = Nodo("[Solución] Reinicia el router y limpia la caché de DNS.")  # No

raiz.der = Nodo("¿El cable de energía está bien conectado?")
raiz.der.izq = Nodo("[Solución] El router podría estar averiado. Contacta a soporte.")  # Sí
raiz.der.der = Nodo("[Solución] Conecta el cable firmemente a la toma eléctrica.")

def preorden(nodo):
    if nodo is not None:
        print(nodo.valor, end=" ")
        preorden(nodo.izq)
        preorden(nodo.der)


def postorden(nodo):
    if nodo is not None:
        postorden(nodo.izq)
        postorden(nodo.der)
        print(nodo.valor, end=" ")

preorden(raiz)
print()
postorden(raiz)
```

### Ejercicio 3: Reporte de Auditoría de Carpetas
Represente una jerarquía de carpetas y archivos mediante un árbol binario donde cada nodo guarda un nombre:

- Use la función del Ejemplo 3 para contar el total de carpetas y archivos almacenados en la estructura.
- Cree una función contar_hojas(nodo) que cuente únicamente los nodos que no tienen hijos a la izquierda ni a la derecha (los archivos finales).
- Aplique la función de búsqueda para verificar si un archivo específico ingresado por el usuario existe dentro del árbol.

```python
def contar_nodos(nodo):
    if nodo is None:
        return 0
    return 1 + contar_nodos(nodo.izq) + contar_nodos(nodo.der)

def contar_hojas(nodo):
    if nodo.izq is None:
        return 1
    if nodo.der is None:
        return 1
    return contar_hojas(nodo.izq) + contar_hojas(nodo.der)


def buscar_archivo(nodo, objetivo):
    if nodo is None:
        return False
    if nodo.valor == objetivo:
        return True
    return buscar_archivo(nodo.izq, objetivo) or buscar_archivo(nodo.der, objetivo)


raiz = Nodo("Raíz")

raiz.izq = Nodo("Documentos")
raiz.der = Nodo("Imágenes")

raiz.izq.izq = Nodo("Reporte.pdf")
raiz.izq.der = Nodo("Proyecto.docx")

raiz.der.izq = Nodo("Foto1.png")
raiz.der.der = Nodo("Banner.jpg")


print(f"Total de nodos (carpetas + archivos): {contar_nodos(raiz)}")
print(f"Total de hojas (archivos finales): {contar_hojas(raiz)}")


if buscar_archivo(raiz, "Foto1.png"):
    print(f"Resultado: El archivo Foto1.png SÍ existe en el sistema.")
else:
    print(f"Resultado: El archivo Foto1.png NO fue encontrado.")

if buscar_archivo(raiz, "Foto2.png"):
    print(f"Resultado: El archivo Foto2.png SÍ existe en el sistema.")
else:
    print(f"Resultado: El archivo Foto2.png NO fue encontrado.")
```
