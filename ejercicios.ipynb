### Ejercicio 1: Árbol Genealógico Familiar
Utilizando la clase Nodo, modele un árbol genealógico de 3 niveles con al menos 7 personas:

- Coloque en la raíz al abuelo/abuela.
- Asigne en los enlaces izquierdo y derecho a dos hijos (padres/tíos).
- Asigne a cada hijo al menos dos descendientes (nietos) en sus respectivos enlaces izquierdo y derecho.
- Imprima las relaciones de forma manual accediendo a los atributos para comprobar que cada nodo apunta al familiar correcto.

### Ejercicio 2: Menú Interactivo de Decisiones
Cree un árbol de toma de decisiones para un sistema de soporte técnico o menú de opciones:

- Construya un árbol donde la raíz sea una pregunta inicial y los enlaces izquierdo y derecho lleven a opciones como "Sí" y "No" con sus respectivas acciones finales.
- Ejecute e imprima el recorrido en Preorden para listar el orden en el que se evaluarían las opciones desde la raíz hacia las hojas.
- Ejecute e imprima el recorrido en Postorden para simular el cierre de los procesos de soporte desde las hojas hacia la raíz.

### Ejercicio 3: Reporte de Auditoría de Carpetas
Represente una jerarquía de carpetas y archivos mediante un árbol binario donde cada nodo guarda un nombre:

- Use la función del Ejemplo 3 para contar el total de carpetas y archivos almacenados en la estructura.
- Cree una función contar_hojas(nodo) que cuente únicamente los nodos que no tienen hijos a la izquierda ni a la derecha (los archivos finales).
- Aplique la función de búsqueda para verificar si un archivo específico ingresado por el usuario existe dentro del árbol.
