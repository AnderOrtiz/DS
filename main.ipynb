```python
class PistaNodoDoble:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None
        self.anterior = None


class ListaDoblementeEnlazada:
    def __init__(self):
        self.cabeza = None
        self.cola = None

    def insertar_inicio(self, dato):
        nuevo_nodo = PistaNodoDoble(dato)
        if self.cola is None:
            self.cola = nuevo_nodo
            self.cabeza = nuevo_nodo
        else:
            nuevo_nodo.siguiente = self.cabeza
            self.cabeza.anterior = nuevo_nodo
            self.cabeza = nuevo_nodo

    def insertar_inicio(self, dato):
        nuevo_nodo = PistaNodoDoble(dato)
        if self.cola is None:
            self.cola = nuevo_nodo
            self.cabeza = nuevo_nodo
        else:
            nuevo_nodo.siguiente = self.cabeza
            self.cabeza.anterior = nuevo_nodo
            self.cabeza = nuevo_nodo

    def insertar_final(self, dato):
        nuevo_nodo = PistaNodoDoble(dato)
        if self.cabeza is None:
            self.cabeza = nuevo_nodo
            self.cola = nuevo_nodo
        else:
            nuevo_nodo.anterior = self.cola
            self.cola.siguiente = nuevo_nodo
            self.cola = nuevo_nodo

    def recorrer_adelante(self):
        actual = self.cabeza
        while actual is not None:
            print(actual.dato, end=" <-> ")
            actual = actual.siguiente
        print("None")

    def recorrer_atras(self):
        actual = self.cola
        while actual is not None:
            print(actual.dato, end=" <-> ")
            actual = actual.anterior
        print("None")


    def buscar_desde_cola(self, objetivo):
        actual = self.cola
        contador = 1

        while actual is not None:
            if actual.dato != objetivo :
                contador += 1

            if actual.dato == objetivo:
                contador_found = contador

            actual = actual.anterior
        print(f" Número de iteraciones buscando desde la cola  hacia la cabeza: {contador_found}")
        print(f" Número de iteraciones buscando desde la cabeza  hacia la cola: {contador - contador_found + 1}")

    def eliminar_nodo(self, objetivo):
        actual = self.cabeza
        while actual is not None and actual.dato != objetivo:
            actual = actual.siguiente

        if actual is None:
            return False

        if actual.anterior is not None:
            actual.anterior.siguiente = actual.siguiente
        else:
            self.cabeza = actual.siguiente

        if actual.siguiente is not None:
            actual.siguiente.anterior = actual.anterior
        else:
            self.cola = actual.anterior

        return True
    
    def eliminar_pista(self, pistas):
        self.eliminar_nodo(pistas)
```
