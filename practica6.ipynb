### 5.1 Mapeo de registros

Inventario donde las claves son códigos SKU y los valores son diccionarios internos con atributos

```python
inventario_almacen = {
    "SKU-101": {"nombre": "Teclado Mecánico RGB", "precio": 52.00, "existencia": 18},
    "SKU-102": {"nombre": "Mouse Ergonómico", "precio": 24.50, "existencia": 40},
    "SKU-103": {"nombre": "Monitor Gamer 27\"", "precio": 210.00, "existencia": 8},
    "SKU-104": {"nombre": "Audífonos Bluetooth", "precio": 45.00, "existencia": 25}
}

print("=== ESTADO INICIAL DEL INVENTARIO ===")
for codigo_sku, datos_producto in inventario_almacen.items():
    nombre_prod = datos_producto["nombre"]
    precio_prod = datos_producto["precio"]
    stock_prod = datos_producto["existencia"]
    print(f"Código: {codigo_sku:<8} | Producto: {nombre_prod:<22} | Precio: ${precio_prod:<6.2f} | Stock: {stock_prod}")

inventario_almacen["SKU-103"]["existencia"] -= 3
inventario_almacen["SKU-105"] = {"nombre": "Cámara Web Full HD", "precio": 38.90, "existencia": 14}

print("\n=== ACTUALIZACIONES REGISTRADAS ===")
print(f"Stock actualizado de SKU-103: {inventario_almacen['SKU-103']['existencia']} unidades")
print(f"Nuevo ítem registrado (SKU-105): {inventario_almacen['SKU-105']['nombre']} (Stock: {inventario_almacen['SKU-105']['existencia']})\n")
```

### 5.2 Contador de ocurrencias

Frecuencia de cada elemento en una lista de eventos usando `.get()` con valores por defecto

```python
registro_transacciones = [
    "COMPRA", "ERROR_PAGO", "CONSULTA", "COMPRA", "COMPRA", 
    "ERROR_PAGO", "CANCELACION", "CONSULTA", "COMPRA"
]

frecuencia_transacciones = {}

for evento_registrado in registro_transacciones:
    frecuencia_transacciones[evento_registrado] = frecuencia_transacciones.get(evento_registrado, 0) + 1

print("=== REGISTRO DE FRECUENCIA DE EVENTOS ===")
for tipo_evento, total_ocurrencias in frecuencia_transacciones.items():
    print(f"Evento: {tipo_evento:<12} | Total de apariciones: {total_ocurrencias}")
print()
```

### 5.3 Agrupamiento por categoría

Elementos organizados en un diccionario según su categoría

```python
listado_inventario_general = [
    ("Teclado Mecánico RGB", "Periféricos"),
    ("Monitor Gamer 27\"", "Pantallas"),
    ("Mouse Ergonómico", "Periféricos"),
    ("Monitor Portátil 15.6\"", "Pantallas"),
    ("Audífonos Bluetooth", "Audio"),
    ("Barra de Sonido Desktop", "Audio"),
    ("Cámara Web Full HD", "Periféricos")
]

agrupado_por_categoria = {}

for item_nombre, categoria_asignada in listado_inventario_general:
    if categoria_asignada not in agrupado_por_categoria:
        agrupado_por_categoria[categoria_asignada] = []
    agrupado_por_categoria[categoria_asignada].append(item_nombre)

print("=== AGRUPAMIENTO POR CATEGORÍA DE PRODUCTOS ===")
for categoria_nombre, lista_items in agrupado_por_categoria.items():
    items_formateados = ", ".join(lista_items)
    print(f"Categoría [{categoria_nombre}]:")
    print(f"  -> Productos: {items_formateados}")
```
